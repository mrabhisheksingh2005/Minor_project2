```python
# ================================================================
# 🌱 AGRIVISION AI - WEED RECOGNITION
# Dataset: DeepWeeds
# Model: MobileNetV2
# Platform: Google Colab
#
# OUTPUT:
# 1. agrivision_weed_best.keras
# 2. agrivision_weed_best.tflite
# 3. weed_labels.txt
# ================================================================


# ================================================================
# 1. INSTALL
# ================================================================

!pip install -q kagglehub scikit-learn


# ================================================================
# 2. IMPORTS
# ================================================================

import os
import glob
import random
import shutil
import numpy as np
import pandas as pd
import tensorflow as tf
import matplotlib.pyplot as plt
import kagglehub

from tensorflow.keras import layers
from sklearn.model_selection import train_test_split


print("=" * 70)
print("🌱 AGRIVISION AI - DEEPWEEDS RECOGNITION")
print("=" * 70)

print("\nTensorFlow:", tf.__version__)


# ================================================================
# 3. CHECK GPU
# ================================================================

gpus = tf.config.list_physical_devices("GPU")

if gpus:
    print("✅ GPU detected:", gpus)
else:
    print("⚠️ GPU NOT DETECTED")
    print("Colab → Runtime → Change runtime type → T4 GPU")


# ================================================================
# 4. SETTINGS
# ================================================================

SEED = 42

IMG_SIZE = 160
BATCH_SIZE = 64

INITIAL_EPOCHS = 8
FINE_TUNE_EPOCHS = 7

AUTOTUNE = tf.data.AUTOTUNE

random.seed(SEED)
np.random.seed(SEED)
tf.random.set_seed(SEED)


print("\n⚙️ SETTINGS")
print("Image size       :", IMG_SIZE)
print("Batch size       :", BATCH_SIZE)
print("Initial epochs   :", INITIAL_EPOCHS)
print("Fine-tune epochs :", FINE_TUNE_EPOCHS)


# ================================================================
# 5. DOWNLOAD DEEPWEEDS
# ================================================================

print("\n")
print("=" * 70)
print("📥 DOWNLOADING DEEPWEEDS")
print("=" * 70)

# Kaggle mirror of DeepWeeds
dataset_path = kagglehub.dataset_download(
    "imsparsh/deepweeds"
)

print("\n✅ Dataset downloaded!")
print("📁 Path:", dataset_path)


# ================================================================
# 6. INSPECT DATASET
# ================================================================

print("\n📂 Dataset files:")

for root, dirs, files in os.walk(dataset_path):

    level = root.replace(
        dataset_path,
        ""
    ).count(os.sep)

    if level <= 2:

        indent = "  " * level

        print(
            indent +
            os.path.basename(root) +
            "/"
        )

        for file in files[:5]:

            print(
                indent +
                "  " +
                file
            )


# ================================================================
# 7. FIND CSV
# ================================================================

csv_files = glob.glob(

    os.path.join(
        dataset_path,
        "**",
        "*.csv"
    ),

    recursive=True

)


print("\n📄 CSV files found:")

for file in csv_files:
    print(file)


# ================================================================
# 8. FIND IMAGES
# ================================================================

extensions = (
    "*.jpg",
    "*.jpeg",
    "*.png",
    "*.JPG",
    "*.JPEG",
    "*.PNG"
)


all_images = []


for extension in extensions:

    all_images.extend(

        glob.glob(

            os.path.join(
                dataset_path,
                "**",
                extension
            ),

            recursive=True

        )

    )


print(
    "\n🖼️ Images found:",
    len(all_images)
)


if len(all_images) == 0:

    raise RuntimeError(
        "❌ No images found!"
    )


# ================================================================
# 9. CREATE IMAGE LOOKUP
# ================================================================

image_lookup = {}


for path in all_images:

    filename = os.path.basename(path)

    image_lookup[filename] = path


print(
    "✅ Image lookup created:",
    len(image_lookup)
)


# ================================================================
# 10. LOAD LABEL INFORMATION
# ================================================================

# DeepWeeds releases commonly contain a labels CSV.
# We automatically inspect CSV files and find one containing
# image filenames and labels.


label_df = None


for csv_path in csv_files:

    try:

        temp = pd.read_csv(
            csv_path
        )

        print(
            "\nChecking:",
            csv_path
        )

        print(
            "Columns:",
            temp.columns.tolist()
        )

        if len(temp) > 100:

            label_df = temp

            print(
                "✅ Using label file:",
                csv_path
            )

            break

    except Exception as e:

        pass


# ================================================================
# 11. FALLBACK - FOLDER BASED DATASET
# ================================================================

folder_based = False


if label_df is None:

    print(
        "\n⚠️ Suitable CSV not detected."
    )

    print(
        "Trying folder-based labels..."
    )

    folder_based = True


# ================================================================
# 12A. CSV BASED LABEL DETECTION
# ================================================================

if not folder_based:

    # Normalize column names

    original_columns = (
        label_df.columns.tolist()
    )

    label_df.columns = [

        str(c).strip().lower()

        for c in label_df.columns

    ]


    print(
        "\nNormalized columns:",
        label_df.columns.tolist()
    )


    # ------------------------------------------------------------
    # Find filename column
    # ------------------------------------------------------------

    filename_column = None


    filename_candidates = [

        "filename",
        "file",
        "image",
        "image_name",
        "imagename",
        "name"

    ]


    for candidate in filename_candidates:

        if candidate in label_df.columns:

            filename_column = candidate

            break


    # ------------------------------------------------------------
    # Find label column
    # ------------------------------------------------------------

    label_column = None


    label_candidates = [

        "label",
        "class",
        "species",
        "category",
        "weed",
        "weed_type"

    ]


    for candidate in label_candidates:

        if candidate in label_df.columns:

            label_column = candidate

            break


    # ------------------------------------------------------------
    # If columns weren't obvious, inspect first columns
    # ------------------------------------------------------------

    if filename_column is None:

        for column in label_df.columns:

            sample_values = (
                label_df[column]
                .astype(str)
                .head(50)
                .tolist()
            )

            image_matches = sum(

                value.lower().endswith(
                    (
                        ".jpg",
                        ".jpeg",
                        ".png"
                    )
                )

                for value in sample_values

            )

            if image_matches > 5:

                filename_column = column

                break


    print(
        "\nFilename column:",
        filename_column
    )

    print(
        "Label column:",
        label_column
    )


# ================================================================
# 13. BUILD PATH + LABEL LISTS
# ================================================================

paths = []
raw_labels = []


if not folder_based and filename_column is not None and label_column is not None:

    print(
        "\n📊 Building dataset from CSV..."
    )


    for _, row in label_df.iterrows():

        filename = os.path.basename(
            str(
                row[filename_column]
            )
        )

        path = image_lookup.get(
            filename
        )


        if path is not None:

            paths.append(
                path
            )

            raw_labels.append(
                str(
                    row[label_column]
                ).strip()
            )


# ================================================================
# 14. FALLBACK TO DIRECTORY NAMES
# ================================================================

if len(paths) == 0:

    print(
        "\n📁 Using folder names as labels..."
    )


    paths = []

    raw_labels = []


    for path in all_images:

        parent = os.path.basename(
            os.path.dirname(path)
        )

        paths.append(
            path
        )

        raw_labels.append(
            parent
        )


print(
    "\nImages with labels:",
    len(paths)
)


# ================================================================
# 15. ENCODE LABELS
# ================================================================

unique_labels = sorted(
    list(
        set(raw_labels)
    )
)


print(
    "\n🌱 Detected classes:"
)


for label in unique_labels:

    print(
        "•",
        label
    )


NUM_CLASSES = len(
    unique_labels
)


print(
    "\nTotal classes:",
    NUM_CLASSES
)


if NUM_CLASSES < 2:

    raise RuntimeError(
        "❌ Could not detect multiple weed classes."
    )


label_to_index = {

    label: index

    for index, label in enumerate(
        unique_labels
    )

}


labels = np.array(

    [

        label_to_index[label]

        for label in raw_labels

    ],

    dtype=np.int32

)


paths = np.array(
    paths
)


# ================================================================
# 16. CLASS DISTRIBUTION
# ================================================================

print("\n📊 CLASS DISTRIBUTION")


for class_name in unique_labels:

    count = raw_labels.count(
        class_name
    )

    print(
        f"{class_name:30s} : {count}"
    )


# ================================================================
# 17. TRAIN / VALIDATION / TEST SPLIT
# ================================================================

# 80% train
# 10% validation
# 10% test


train_paths, temp_paths, train_labels, temp_labels = train_test_split(

    paths,

    labels,

    test_size=0.20,

    random_state=SEED,

    stratify=labels

)


val_paths, test_paths, val_labels, test_labels = train_test_split(

    temp_paths,

    temp_labels,

    test_size=0.50,

    random_state=SEED,

    stratify=temp_labels

)


print("\n")
print("=" * 70)
print("📊 DATASET SPLIT")
print("=" * 70)


print(
    "Training   :",
    len(train_paths)
)

print(
    "Validation :",
    len(val_paths)
)

print(
    "Testing    :",
    len(test_paths)
)


# ================================================================
# 18. IMAGE PREPROCESSING
# ================================================================

def load_image(
    path,
    label
):

    image = tf.io.read_file(
        path
    )


    image = tf.io.decode_image(

        image,

        channels=3,

        expand_animations=False

    )


    image.set_shape(
        [None, None, 3]
    )


    image = tf.image.resize(

        image,

        [
            IMG_SIZE,
            IMG_SIZE
        ]

    )


    image = (
        tf.keras.applications
        .mobilenet_v2
        .preprocess_input(
            image
        )
    )


    return image, label


# ================================================================
# 19. DATASET CREATOR
# ================================================================

def create_dataset(
    paths,
    labels,
    training=False
):


    ds = (
        tf.data.Dataset
        .from_tensor_slices(
            (
                paths,
                labels
            )
        )
    )


    if training:

        ds = ds.shuffle(

            min(
                len(paths),
                10000
            ),

            seed=SEED

        )


    ds = ds.map(

        load_image,

        num_parallel_calls=AUTOTUNE

    )


    ds = ds.batch(
        BATCH_SIZE
    )


    ds = ds.prefetch(
        AUTOTUNE
    )


    return ds


# ================================================================
# 20. CREATE DATASETS
# ================================================================

train_ds = create_dataset(

    train_paths,

    train_labels,

    True

)


val_ds = create_dataset(

    val_paths,

    val_labels

)


test_ds = create_dataset(

    test_paths,

    test_labels

)


print(
    "\n✅ TensorFlow datasets ready!"
)


# ================================================================
# 21. DATA AUGMENTATION
# ================================================================

augmentation = tf.keras.Sequential(

    [

        layers.RandomFlip(
            "horizontal"
        ),

        layers.RandomRotation(
            0.10
        ),

        layers.RandomZoom(
            0.10
        ),

        layers.RandomContrast(
            0.10
        )

    ],

    name="weed_augmentation"

)


# ================================================================
# 22. LOAD MOBILENETV2
# ================================================================

print(
    "\n🧠 Loading MobileNetV2..."
)


base_model = tf.keras.applications.MobileNetV2(

    input_shape=(
        IMG_SIZE,
        IMG_SIZE,
        3
    ),

    include_top=False,

    weights="imagenet"

)


base_model.trainable = False


print(
    "✅ MobileNetV2 loaded!"
)


# ================================================================
# 23. BUILD MODEL
# ================================================================

inputs = tf.keras.Input(

    shape=(
        IMG_SIZE,
        IMG_SIZE,
        3
    )

)


x = augmentation(
    inputs
)


x = base_model(

    x,

    training=False

)


x = layers.GlobalAveragePooling2D()(
    x
)


x = layers.Dropout(
    0.30
)(
    x
)


x = layers.Dense(

    256,

    activation="relu"

)(
    x
)


x = layers.Dropout(
    0.20
)(
    x
)


outputs = layers.Dense(

    NUM_CLASSES,

    activation="softmax"

)(
    x
)


model = tf.keras.Model(

    inputs,

    outputs,

    name="AgriVision_DeepWeeds"

)


# ================================================================
# 24. COMPILE STAGE 1
# ================================================================

model.compile(

    optimizer=tf.keras.optimizers.Adam(
        learning_rate=0.001
    ),

    loss="sparse_categorical_crossentropy",

    metrics=[
        "accuracy"
    ]

)


model.summary()


# ================================================================
# 25. STAGE 1 CALLBACKS
# ================================================================

STAGE1_PATH = (
    "/content/"
    "weed_stage1_best.keras"
)


callbacks1 = [

    tf.keras.callbacks.ModelCheckpoint(

        STAGE1_PATH,

        monitor="val_accuracy",

        mode="max",

        save_best_only=True,

        verbose=1

    ),


    tf.keras.callbacks.EarlyStopping(

        monitor="val_accuracy",

        mode="max",

        patience=3,

        restore_best_weights=True,

        verbose=1

    ),


    tf.keras.callbacks.ReduceLROnPlateau(

        monitor="val_loss",

        factor=0.3,

        patience=2,

        min_lr=1e-6,

        verbose=1

    )

]


# ================================================================
# 26. STAGE 1 TRAINING
# ================================================================

print("\n")
print("=" * 70)
print("🚀 STAGE 1 - TRANSFER LEARNING")
print("=" * 70)


history1 = model.fit(

    train_ds,

    validation_data=val_ds,

    epochs=INITIAL_EPOCHS,

    callbacks=callbacks1

)


# ================================================================
# 27. LOAD BEST STAGE 1
# ================================================================

stage1_model = tf.keras.models.load_model(
    STAGE1_PATH
)


stage1_loss, stage1_accuracy = (
    stage1_model.evaluate(
        test_ds,
        verbose=1
    )
)


print(
    "\n🎯 Stage 1 Test Accuracy:"
)


print(
    f"{stage1_accuracy * 100:.2f}%"
)


# ================================================================
# 28. FINE-TUNING
# ================================================================

print("\n")
print("=" * 70)
print("🔥 STAGE 2 - FINE TUNING")
print("=" * 70)


model = tf.keras.models.load_model(
    STAGE1_PATH
)


# Find MobileNet backbone

base_model = None


for layer in model.layers:

    if isinstance(
        layer,
        tf.keras.Model
    ):

        if "mobilenet" in layer.name.lower():

            base_model = layer

            break


if base_model is None:

    raise RuntimeError(
        "❌ MobileNetV2 backbone not found!"
    )


# ================================================================
# 29. UNFREEZE LAST 40 LAYERS
# ================================================================

base_model.trainable = True


for layer in base_model.layers:

    layer.trainable = False


for layer in base_model.layers[-40:]:

    # Keep BatchNormalization frozen

    if not isinstance(
        layer,
        tf.keras.layers.BatchNormalization
    ):

        layer.trainable = True


trainable_layers = sum(

    1

    for layer in base_model.layers

    if layer.trainable

)


print(
    "🔥 Trainable MobileNet layers:",
    trainable_layers
)


# ================================================================
# 30. RECOMPILE
# ================================================================

model.compile(

    optimizer=tf.keras.optimizers.Adam(
        learning_rate=1e-5
    ),

    loss="sparse_categorical_crossentropy",

    metrics=[
        "accuracy"
    ]

)


# ================================================================
# 31. STAGE 2 CALLBACKS
# ================================================================

STAGE2_PATH = (
    "/content/"
    "weed_finetuned_best.keras"
)


callbacks2 = [

    tf.keras.callbacks.ModelCheckpoint(

        STAGE2_PATH,

        monitor="val_accuracy",

        mode="max",

        save_best_only=True,

        verbose=1

    ),


    tf.keras.callbacks.EarlyStopping(

        monitor="val_accuracy",

        mode="max",

        patience=3,

        restore_best_weights=True,

        verbose=1

    ),


    tf.keras.callbacks.ReduceLROnPlateau(

        monitor="val_loss",

        factor=0.3,

        patience=2,

        min_lr=1e-7,

        verbose=1

    )

]


# ================================================================
# 32. FINE-TUNE MODEL
# ================================================================

history2 = model.fit(

    train_ds,

    validation_data=val_ds,

    epochs=FINE_TUNE_EPOCHS,

    callbacks=callbacks2

)


# ================================================================
# 33. LOAD BEST FINE-TUNED MODEL
# ================================================================

fine_model = tf.keras.models.load_model(
    STAGE2_PATH
)


fine_loss, fine_accuracy = (
    fine_model.evaluate(
        test_ds,
        verbose=1
    )
)


print(
    "\n🎯 Fine-Tuned Test Accuracy:"
)


print(
    f"{fine_accuracy * 100:.2f}%"
)


# ================================================================
# 34. CHOOSE BEST MODEL
# ================================================================

print("\n")
print("=" * 70)
print("🏆 MODEL COMPARISON")
print("=" * 70)


print(
    f"Stage 1      : "
    f"{stage1_accuracy * 100:.2f}%"
)


print(
    f"Fine-tuned   : "
    f"{fine_accuracy * 100:.2f}%"
)


if fine_accuracy >= stage1_accuracy:

    best_model = fine_model

    best_accuracy = fine_accuracy

    best_name = "Fine-tuned MobileNetV2"

else:

    best_model = stage1_model

    best_accuracy = stage1_accuracy

    best_name = "Frozen MobileNetV2"


print(
    "\n🥇 Selected:",
    best_name
)


# ================================================================
# 35. SAVE FINAL KERAS MODEL
# ================================================================

FINAL_KERAS = (
    "/content/"
    "agrivision_weed_best.keras"
)


best_model.save(
    FINAL_KERAS
)


print(
    "\n✅ Keras model saved:"
)

print(
    FINAL_KERAS
)


# ================================================================
# 36. CONVERT TO TFLITE
# ================================================================

print(
    "\n📱 Converting to TFLite..."
)


converter = (
    tf.lite.TFLiteConverter
    .from_keras_model(
        best_model
    )
)


converter.optimizations = [
    tf.lite.Optimize.DEFAULT
]


tflite_model = (
    converter.convert()
)


FINAL_TFLITE = (
    "/content/"
    "agrivision_weed_best.tflite"
)


with open(
    FINAL_TFLITE,
    "wb"
) as f:

    f.write(
        tflite_model
    )


print(
    "✅ TFLite created:"
)

print(
    FINAL_TFLITE
)


# ================================================================
# 37. SAVE LABELS
# ================================================================

LABEL_PATH = (
    "/content/"
    "weed_labels.txt"
)


with open(
    LABEL_PATH,
    "w",
    encoding="utf-8"
) as f:

    for label in unique_labels:

        f.write(
            str(label)
            +
            "\n"
        )


print(
    "✅ Weed labels saved:"
)

print(
    LABEL_PATH
)


# ================================================================
# 38. ACCURACY GRAPH
# ================================================================

train_accuracy = (

    history1.history["accuracy"]

    +

    history2.history["accuracy"]

)


val_accuracy = (

    history1.history["val_accuracy"]

    +

    history2.history["val_accuracy"]

)


plt.figure(
    figsize=(9, 5)
)


plt.plot(
    train_accuracy,
    label="Training Accuracy"
)


plt.plot(
    val_accuracy,
    label="Validation Accuracy"
)


plt.axvline(

    x=len(
        history1.history["accuracy"]
    ) - 1,

    linestyle="--",

    label="Fine-tuning"

)


plt.xlabel(
    "Epoch"
)


plt.ylabel(
    "Accuracy"
)


plt.title(
    "AgriVision AI - DeepWeeds Accuracy"
)


plt.legend()

plt.grid()

plt.show()


# ================================================================
# 39. LOSS GRAPH
# ================================================================

train_loss = (

    history1.history["loss"]

    +

    history2.history["loss"]

)


val_loss = (

    history1.history["val_loss"]

    +

    history2.history["val_loss"]

)


plt.figure(
    figsize=(9, 5)
)


plt.plot(
    train_loss,
    label="Training Loss"
)


plt.plot(
    val_loss,
    label="Validation Loss"
)


plt.axvline(

    x=len(
        history1.history["loss"]
    ) - 1,

    linestyle="--",

    label="Fine-tuning"

)


plt.xlabel(
    "Epoch"
)


plt.ylabel(
    "Loss"
)


plt.title(
    "AgriVision AI - DeepWeeds Loss"
)


plt.legend()

plt.grid()

plt.show()


# ================================================================
# 40. FILE SIZES
# ================================================================

keras_size = (

    os.path.getsize(
        FINAL_KERAS
    )

    / 1024

    / 1024

)


tflite_size = (

    os.path.getsize(
        FINAL_TFLITE
    )

    / 1024

    / 1024

)


# ================================================================
# 41. FINAL SUMMARY
# ================================================================

print("\n")
print("=" * 70)
print("🎉 AGRIVISION AI WEED MODEL COMPLETE")
print("=" * 70)


print(

f"""

DATASET
-------
DeepWeeds

Images:
{len(paths)}

Classes:
{NUM_CLASSES}


MODEL
-----
MobileNetV2

Image Size:
{IMG_SIZE} x {IMG_SIZE}


TRAINING
--------
Stage 1:
Maximum {INITIAL_EPOCHS} epochs

Stage 2:
Maximum {FINE_TUNE_EPOCHS} epochs


RESULTS
-------
Stage 1 Test Accuracy:
{stage1_accuracy * 100:.2f}%

Fine-tuned Test Accuracy:
{fine_accuracy * 100:.2f}%

BEST MODEL:
{best_name}

FINAL TEST ACCURACY:
{best_accuracy * 100:.2f}%


FILES
-----
agrivision_weed_best.keras

agrivision_weed_best.tflite

weed_labels.txt


MODEL SIZE
----------
Keras:
{keras_size:.2f} MB

TFLite:
{tflite_size:.2f} MB


FLUTTER
-------
Use:

agrivision_weed_best.tflite

+

weed_labels.txt

"""

)

print("=" * 70)


# ================================================================
# 42. DOWNLOAD FLUTTER FILES
# ================================================================

from google.colab import files


print(
    "\n📥 Downloading TFLite model..."
)

files.download(
    FINAL_TFLITE
)


print(
    "\n📥 Downloading labels..."
)

files.download(
    LABEL_PATH
)


print(
    "\n💾 Master Keras model remains at:"
)

print(
    FINAL_KERAS
)


print(
    "\n🎉 DONE!"
)
```
